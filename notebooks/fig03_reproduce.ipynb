```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### HCP dataset

```python
lambdas = np.linspace(0.10, 0.90, 41)
res = pd.read_csv('../results/fig03/graphical_res_HCP_MMP(distance_ridge).csv')
filtered_res = res[(res['lambda'] > 0.1-1e-3) & (res['lambda'] < 0.9+1e-3)]
```

```python
summary = filtered_res.drop(columns = ['FILE_ID']).groupby('lambda').agg(['mean', 'std'])['global_cost']

v_mean = summary['mean'].values; v_std = summary['std'].values
plt.plot(lambdas, v_mean, color = 'k', zorder = 2)

for sub in np.unique(filtered_res['FILE_ID'].values):
    sub_temp = np.array(filtered_res[filtered_res['FILE_ID'] == sub]['global_cost'])
    plt.plot(lambdas, sub_temp, color = 'darkgray', alpha = 0.3, zorder = 1)
for l in [0.1, 0.3, 0.5, 0.7, 0.9]:
    plt.scatter(l, v_mean[np.where(abs(lambdas-l)<1e-3)[0][0]], s = 120, zorder = 2, lw = .5, edgecolors = 'white', label = r'$\boldsymbol{W}$ '+r'($\lambda$ = {:.1f})'.format(l))
plt.text(x = 0.470, y = 0.092, s = 'Biological brain', color = 'r', fontsize = 12)

ax = plt.gca(); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
plt.xlabel(r'$\lambda$', fontsize = 12); plt.xlim(0.050, 0.950); plt.xticks([0.1, 0.3, 0.5, 0.7, 0.9], fontsize = 10)
plt.ylabel('Cost (HCP)', fontsize = 12); plt.ylim(0.090, 0.122); plt.yticks([0.09, 0.10, 0.11, 0.12], fontsize = 10)
plt.legend(loc = 'upper right', fontsize = 10)

plt.show()
```

```python
summary = filtered_res.drop(columns = ['FILE_ID']).groupby('lambda').agg(['mean', 'std'])['global_effi']

v_mean = summary['mean'].values; v_std = summary['std'].values
plt.plot(lambdas, v_mean, color = 'k', zorder = 2)

for sub in np.unique(filtered_res['FILE_ID'].values):
    sub_temp = np.array(filtered_res[filtered_res['FILE_ID'] == sub]['global_effi'])
    plt.plot(lambdas, sub_temp, color = 'darkgray', alpha = 0.3, zorder = 1)
for l in [0.1, 0.3, 0.5, 0.7, 0.9]:
    plt.scatter(l, v_mean[np.where(abs(lambdas-l)<1e-3)[0][0]], s = 120, zorder = 2, lw = .5, edgecolors = 'white', label = r'$\boldsymbol{W}$ '+r'($\lambda$ = {:.1f})'.format(l))
plt.text(x = 0.470, y = 0.561, s = 'Biological brain', color = 'r', fontsize = 12)

ax = plt.gca(); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
plt.xlabel(r'$\lambda$', fontsize = 12); plt.xlim(0.050, 0.950); plt.xlim(0.050, 0.950); plt.xticks([0.1, 0.3, 0.5, 0.7, 0.9], fontsize = 10)
plt.ylabel('Efficiency (HCP)', fontsize = 12); plt.ylim(0.545, 0.575); plt.yticks([0.55, 0.56, 0.57], fontsize = 10)
plt.legend(loc = 'lower left', fontsize = 10)

plt.show()
```

```python
plt.scatter(filtered_res['global_cost'], filtered_res['global_effi'], color = 'grey', zorder = 1, s = 10)

for i in range(5):
    l = 0.2 * i + 0.1
    plt.scatter(
        filtered_res[abs(filtered_res['lambda'] - l) < 1e-3]['global_cost'], 
        filtered_res[abs(filtered_res['lambda'] - l) < 1e-3]['global_effi'], 
        color = plt.get_cmap('tab10').colors[i], zorder = 2, s = 10, 
        label = r'$\boldsymbol{W}$ '+r'($\lambda$ = {:.1f})'.format(l)
    )
plt.text(x = 0.098, y = 0.561, s = 'Biological brain', color = 'r', fontsize = 12)

ax = plt.gca(); ax.spines['right'].set_visible(False); ax.spines['top'].set_visible(False)
plt.xlabel('Cost (HCP)', fontsize = 12); plt.xlim(0.085, 0.125); plt.xticks([0.09, 0.10, 0.11, 0.12], fontsize = 10)
plt.ylabel('Efficiency (HCP)', fontsize = 12); plt.ylim(0.544, 0.576); plt.yticks([0.55, 0.56, 0.57], fontsize = 10)
plt.legend(loc = 'lower right', fontsize = 10)

plt.show()
```
